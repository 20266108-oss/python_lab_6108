```python
numbers = range(1, 101)  # 1부터 100까지의 숫자를 포함하는 범위를 생성합니다.
sum_numbers = sum(numbers)  # 범위 내 모든 숫자의 합계를 계산합니다.
count_numbers = len(numbers)  # 범위 내 숫자의 개수를 계산합니다.
average = sum_numbers / count_numbers  # 합계를 개수로 나누어 평균을 계산합니다.

print(f"The sum of numbers from 1 to 100 is: {sum_numbers}")  # 합계 출력
print(f"The count of numbers from 1 to 100 is: {count_numbers}")  # 개수 출력
print(f"The average of numbers from 1 to 100 is: {average}")  # 평균 출력
```

코딩 블록에서 사용된 함수에 대한 설명입니다:

*   **`range(start, stop)`**: `start`부터 `stop-1`까지의 정수 시퀀스를 생성합니다. 이 경우 `range(1, 101)`은 1부터 100까지의 숫자를 생성합니다.
*   **`sum(iterable)`**: `iterable` (예: 리스트, 튜플 또는 `range` 객체)에 있는 항목들의 합계를 반환합니다.
*   **`len(s)`**: 객체 `s` (예: 문자열, 바이트, 튜플, 리스트 또는 `range` 객체)의 길이(항목 수)를 반환합니다.

### **변수와** 변수형

```python
# 1. 정수형 (Integer) - 소수점이 없는 숫자
a = 123
print(type(a))  # 결과: <class 'int'>

# 2. 실수형 (Float) - 소수점이 있는 숫자
a = 123.456
print(type(a))  # 결과: <class 'float'>

# 3. 문자열 (String) - 텍스트 데이터 (따옴표로 둘러싸임)
a = "Hello World!"
print(type(a))  # 결과: <class 'str'>

# 4. 논리형/불리언 (Boolean) - 참(True) 또는 거짓(False)
a = True
print(type(a))  # 결과: <class 'bool'>

# 5. 리스트 (List) - 여러 개의 값을 순서대로 담는 배열
a = [1, 2, 3]
print(type(a))  # 결과: <class 'list'>
```

type() 함수: 괄호 안에 있는 변수나 데이터가 어떤 형태(클래스)인지 확인해 주는 내장 함수입니다. 데이터를 다루다가 이 변수가 숫자인지 문자인지 헷갈릴 때 유용하게 쓰입니다.

자유로운 변수 재할당: C나 Java 같은 언어와 달리, 파이썬은 처음 a에 정수(123)를 넣었더라도 나중에 문자열("Hello World!")이나 리스트([1, 2, 3])로 자유롭게 바꿔서 저장할 수 있습니다. 이를 동적 타이핑이라고 부릅니다.**굵은 텍스트**

```python
# 1. 불리언의 숫자 변환 및 연산
# 파이썬에서 True는 내부적으로 1, False는 0으로 처리됩니다.
a = True; b = False
print(a + b)  # 결과: 1 (1 + 0 = 1)

# 2. 산술 연산자 (+)
# 변수 a와 b의 값을 더하여 새로운 변수 c에 저장합니다.
a = 3; b = 4
c = a + b
print(c)  # 결과: 7

# 3. 비교 연산자 (<)
# a가 b보다 작은지 비교합니다. 조건이 맞으므로 불리언 값(True)이 저장됩니다.
d = a < b
print(d)  # 결과: True

# 4. 논리 연산자 (and)
# 두 비교 연산(3 < 4, 4 < 5)이 모두 True일 때만 최종적으로 True가 출력됩니다.
e = 3 < 4 and 4 < 5
print(e)  # 결과: True
```

불리언의 수치화 (a + b): 파이썬에서 True는 숫자 1, False는 숫자 0과 완전히 같습니다. 따라서 산술 연산을 수행하면 1 + 0으로 계산되어 1이 출력됩니다.

비교 연산의 결과 (a < b): 두 값을 비교하는 연산자(<, >, == 등)는 계산 결과로 항상 불리언 값(True 또는 False)을 반환합니다.

논리 연산자 and: A and B 구조에서 A와 B가 모두 참(True)일 때만 True를 반환합니다. 3 < 4(참)와 4 < 5(참)가 모두 성립하므로 결과는 True가 됩니다

## **리스트**
- 여러개의 값을 다룰 때 사용하며 전체 원소를 대괄호([ ])로 묶고, 각 원소는 쉼표(,)로 구분
- 리스트는 모든 데이터 타입의 값을 저장 가능하며, 리스트도 원소로 저장 가능



```python
# 5개의 숫자를 가진 리스트를 생성합니다.
a = [1, 2, 3, 4, 5]

# 1. 인덱싱 (Indexing) - 특정 위치의 값 가져오기
# 파이썬은 순서를 0부터 세기 때문에, a[2]는 세 번째 요소인 '3'을 의미합니다.
b = a[2]
print(b)  # 결과: 3

# 2. 리스트에 요소 추가하기 (append)
# append() 메서드는 리스트의 맨 마지막에 새로운 값을 덧붙입니다.
a.append(6)
print(a)  # 결과: [1, 2, 3, 4, 5, 6]

# 3. 리스트의 특정 요소 수정하기
# 세 번째 요소(인덱스 2)의 값을 기존의 3에서 7로 덮어씁니다.
a[2] = 7
print(a)  # 결과: [1, 2, 7, 4, 5, 6]
```

append() 메서드: 기존 리스트를 유지한 채 맨 끝에 딱 하나의 값을 추가할 때 사용하는 가장 흔한 리스트 조작 방법입니다.

### **튜플**

- 리스트와 유사하나 원소를 추가나 삭제, 변경이 불가능함
- 각 원소를 소괄호(( ))로 묶고, 각 원소는 쉼표(,)로 구분

```python
# 1. 튜플 생성 및 인덱싱 (Indexing)
# 소괄호()를 사용하여 튜플을 만듭니다. 리스트처럼 인덱스(0부터 시작)로 접근할 수 있습니다.
a = (1, 2, 3, 4, 5)
b = a[2]  # 세 번째 위치(인덱스 2)의 값인 3을 가져옵니다.
print(b)  # 결과: 3

# 2. 리스트 언패킹 (List Unpacking)
# 리스트 안의 요소들을 순서대로 각각의 변수(a1, a2, a3)에 나누어 담습니다.
a = [1, 2, 3]
a1, a2, a3 = a  # a1에는 1, a2에는 2, a3에는 3이 저장됩니다.
print(a1, a2, a3)  # 결과: 1 2 3

# 3. 튜플 언패킹 (Tuple Unpacking)
# 튜플 역시 리스트와 마찬가지로 언패킹이 가능합니다.
b = (4, 5, 6)
b1, b2, b3, = b  # b1에는 4, b2에는 5, b3에는 6이 저장됩니다. (b3 뒤의 쉼표는 파이썬에서 무시됨)
print(b1, b2, b3)  # 결과: 4 5 6
```

튜플(Tuple)의 성질: 튜플은 소괄호 ()를 사용해 만들며, 리스트와 달리 한 번 생성하면 내부 요소를 수정하거나 삭제할 수 없습니다. 하지만 값을 읽어오는 작업(a[2]와 같은 인덱싱)은 리스트와 동일하게 수행할 수 있습니다.

언패킹(Unpacking): 담겨 있는 묶음 데이터(리스트, 튜플 등)를 풀어서 왼쪽에 있는 변수들에 1:1로 대응시켜 한 번에 저장하는 기술입니다.

변수의 개수와 묶음 내부 요소의 개수가 정확히 일치해야 합니다. (개수가 다르면 ValueError가 발생합니다.)

문법 팁 (b3,): 마지막 변수 뒤에 쉼표(,)가 붙어있어도 파이썬이 문법 오류로 처리하지 않고 정상적으로 값을 할당해 줍니다.

### **딕셔너리**
키(key)와 값(value)을 조합하여 저장 가능.

```python
# 1. 딕셔너리 생성
# 중괄호 {}를 사용해 'Key: Value' (키: 값) 쌍 형태로 데이터를 저장합니다.
a = {"Apple": 300, "Pineapple": 400}

# 2. 딕셔너리 값 조회
# 대괄호 [] 안에 '키(Key)'를 입력하면 대응하는 '값(Value)'을 가져옵니다.
print(a["Apple"])  # 결과: 300

# 3. 원소 값 변경 (수정)
# 이미 존재하는 키("Pineapple")에 새로운 값을 대입하면 기존 값이 덮어씌워집니다.
a["Pineapple"] = 450
print(a["Pineapple"])  # 결과: 450

# 4. 새 원소 추가
# 존재하지 않는 새로운 키("Melon")에 값을 대입하면 딕셔너리에 새 쌍이 추가됩니다.
a["Melon"] = 500
print(a)  # 결과: {'Apple': 300, 'Pineapple': 450, 'Melon': 500}
```

키(Key)와 값(Value)의 구조: 사전에서 단어를 찾으면 뜻이 나오듯, 딕셔너리는 키를 통해 값을 빠르게 찾아내는 구조입니다. (예: "Apple"이라는 키를 열면 300이라는 값이 나옵니다.)

유연한 수정과 추가:

a[키] = 값 형태로 문법은 동일하지만, 존재하는 키면 값이 수정되고 없는 키면 새로 추가됩니다.

순서가 아닌 키로 접근: 리스트는 a[0], a[1]처럼 숫자 인덱스로 접근하지만, 딕셔너리는 a["Apple"]처럼 우리가 지정한 키 이름을 사용해서 접근한다는 점이 가장 큰 차이점입니다.
